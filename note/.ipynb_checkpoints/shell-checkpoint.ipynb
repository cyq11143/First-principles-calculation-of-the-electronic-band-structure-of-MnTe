# 这是一个关于linux系统与shell脚本命令的笔记

## 1. linux常见命令

* ls：显示当前路径所有文件和文件夹名称<br>
ls -l：也可写作ll，作用是列出各文件和文件夹的详细信息(具体可参考https://www.cnblogs.com/CrispyCandy/p/17586142.html)
* pwd：显示当前所在的绝对路径
* history：显示历史输入命令
* cd：进入某文件夹
cd后面不加地址，执行时默认进入/home/用户名文件夹(当前用户身份对应的家目录)<br>
cd -：返回上一次所在目录
* cp：复制文件或者文件夹<br>
cp A B：复制名称为A的文件，新文件名或者路径为B<br>
cp A ~/：把A文件复制到家目录(家目录路径用该简便表示法)<br>
cp -r A B：复制名称为A的文件夹，新文件夹名为B<br>
cp A B C：把A、B文件复制到C<font color=red>文件夹</font><br>
* mv：剪切文件或者文件夹(与cp用法相似)
* 相对路径表示法："."表示当前路径，"../"表示上一级路径
* touch：创建一个文本文件
* mkdir：创建一个文件夹
* rm：删除文件或文件夹<br>
rm A：删除文件A<br>
rm -r A：文件夹A<br>
rm [\^IKPs]*：<font color=red>反向删除</font>，排除掉方括号中包含字母开头的文件，其余全部删除(非常便捷)<br>
* vim：文本编辑<br>
只读模式：vim A<br>
编辑模式：i<br>
退出编辑模式：Esc键<br>
退出文件(不保存修改)：":q"或者":!q"<br>
退出文件(保存修改)：":wq"<br>
单行复制：光标移动至目标行，按yy复制；光标移动至另一行，按p粘贴至<font color=red>下一行</font><br>
多列复制：先ctrl+v，移动光标，选择区域，按y复制；移动光标至某一位置，按下p后粘贴至右方<br>
单行剪切：光标移动至目标行，按dd剪切；光标移动至另一行，按p粘贴至<font color=red>下一行</font><br>
多列剪切：先ctrl+v，移动光标，选择区域，按d剪切；移动光标至某一位置，按下p后粘贴至右方<br>
多行复制/剪切：按下nyy/npp后选择区域，移动光标至某一位置，按p粘贴<br>
D/dd：整行删除<br>
<img src="./vim.png" width="40%"><br>
* grep：在文件中找出含某一字符串的行数<br>
grep str file：在file中找出所有含str字符串的行<br>
grep -An str file：在file中找出所有含str字符串的行及其后n行(after)<br>
grep -Bn str file：在file中找出所有含str字符串的行及其前n行(before)<br>
* more/less：浏览文件，其中more只能向下翻，less可以上下翻阅；浏览完毕时按q即可退出<br>
* cat：将文件内容输出到屏幕(相比more/less的区别是，cat直接把文件所有内容全部输出到屏幕，而more/less则为阅读模式，仅仅显示部分内容)<br>
cat -n A：显示A文件每行的行数<br>
cat A B > C：将文件A、B的内容写入C中<br>
* head：输出文件的头几行(默认10行)<br>
head -n num file：输出file头num行<br>
* tail：输出文件的末几行(默认10行)<br>
tail -n num file：输出file末num行<br>
tail -f file：实时输出最后10行<br>
* split：把一个文件分成许多小文件<br>
split -l n file：以n行为一组，把file分成诸多份小文件<br>
* echo：类似于cat，也是输出文本内容<br>
echo A >> B：把内容A写入B中<br>
* find：找某个文件的路径<br>
find path -name str：在路径path下的所有子目录中找到名称为str的文件的路径<br>
* 特殊符号：有?和*两种<br>
?：表示单个字符，例如：文件xa1、xa2可写作xa?；fit、fat可写作???<br>
*：表示任意长度的字符，例如：文件data、dog可写作d\*<br>
* 输出重定向：>和>>(前者是完全替代原有内容，后者是追加填入内容)<br>
* diff：比较两个文件是否一样(可用于检查某文件是否已经复制过)<br>
diff A B：比较A、B是否一样<br>
diff -y A B：把A、B不同的地方并列出来<br>
* 管道命令："|"，把管道前命令输出的内容重新当成一个文件，使用管道后的命令进行输出<br>
例如：head -5 file | tail -1：该命令表示将file文件前5行当成新文件B，之后执行输出B文件最后一行的命令(等价于输出A文件第五行)<br>

## 2. shell脚本常见命令以及程序语法

* \\$：该字符有许多功能，包括引用变量、位置参数、命令输出、表达式求值等<br>
\\$a：表示引用变量a的值<br>
\\${a}：一样表示引用变量a的值，相比\\$a的优点在于它可以嵌在文本中<br>
\\$n：表示引用内容的第n个参数<br>
\\$0：输出所有引用内容<br>
\\$((a+b))：用变量a和b的值进行代数运算，给出结果<br>
* awk：逐行提取、处理数据<br>
基本格式为：awk '{command}' file，其中command为每行进行的操作命令，file为读取的文件；也可用于管道命令：file | awk {command}<br>
awk '{print \\$2}' File：打印file每一行的第二列<br>
awk '{print NR, NF}' File：打印每一行的行序号，总列数(即该行的总字数)，NR、NF是内置的迭代变量<br>
awk -F,'{command}' File：以","为分隔符(默认以空格为分割符)，-F关键词用于设置分隔符<br>
* sed：处理、编辑文本文件<br>
sed -e 's/str1/str2' file：'-e'表示执行后面的脚本内容，'s/'表示替换，整个命令表示将file中的<font color=red>第一个</font>str1替换成str2，并输出到屏幕(<font color=red>不改变</font>原文件)<br>
sed -e 's/str1/str2/g' file：'/g'表示全局，整个命令表示将file中的<font color=red>所有</font>的str1替换成str2，并输出到屏幕(<font color=red>不改变</font>原文件)<br>
sed -i 's/str1/str2/g' file：'-i'表示<font color=red>修改</font>原文件，整个命令表示将file中的所有的str1替换成str2<br>
更多用法参考https://www.runoob.com/linux/linux-comm-sed.html<br>
* 循环语句：有for循环，格式如下：<br>
for i in list(e.g. list=1 2 3 4或者'seq 1 4')<br>
do<br>
...<br>
done<br>
* 条件判断：if语句，格式如下：<br>
if [command]; then<br>
...<br>
elif [command]; then<br>
...<br>
else<br>
...<br>
fi<br>
* 数量关系符号：多用于if语句<br>
-eq：等于<br>
-ne：不等于<br>
-lt：小于<br>
-le：小于等于<br>
-gt：大于<br>
-ge：大于等于<br>
* 逻辑符号<br>
逻辑与：&&<br>
逻辑或：||<br>
字符串判断：等于(==)、不等于(!=)<br>
